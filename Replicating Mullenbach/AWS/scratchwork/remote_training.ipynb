```python
from sagemaker.pytorch import PyTorch
import time
import sagemaker
```

```python
sagemaker_session = sagemaker.Session()
role = sagemaker.get_execution_role()
```

```python
estimator = PyTorch(
    entry_point = "train.py",
    role        = role,
    py_version  = "py39",
    framework_version="1.13.1",
    instance_count  = 1,
    instance_type   = "ml.m5.large", #"ml.c5.2xlarge",
    hyperparameters = {"lr": 0.01, "num_epochs":10},
)
```

```python
start = time.time()
estimator.fit()
end   = time.time()
print(f'Runtime: {end - start:,.2f}s')
```
